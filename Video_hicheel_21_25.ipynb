###Recursion

n = 5
fac(n)=1*2*3*...*(n-1)*n
fac(5)=1*2*3*4*5
fac(3)=1*2*3

```python
def factorial(n):
  if n == 1 or n == 0:
     return 1
  else:
      return n * factorial(n-1)
```

```python
factorial(5)
```

```python
factorial(6)
```

###Filter

####filter(function,squence)

```python
from string import ascii_lowercase,ascii_uppercase
```

```python
print(ascii_lowercase,ascii_uppercase)
```

```python
def solve(s):
  for c in s:
    if c in ascii_lowercase or c in ascii_uppercase or c == ' ':
      return True
    else:
      return False
```

```python
"".join(list(filter(solve, 'The !notebook wep #41 sok.')))
```

```python
list(filter(lambda x: x % 2 == 0, [11, 20, 30, 49, 99, 100]))
```

```python
"".join(list(filter(lambda s:s.lower()=='!', 'The !notebook dafok peoprothe[elf[hagiugjkljkljdahashdljhaldkjhfhqfh;dioufgwep #41 sok.')))
```

```python
list(filter(lambda x: x % 11 == 0, [11, 20, 30, 49, 99, 100]))
```

###Reduce

```python
import functools
lst = [1, 2, 3, 4, 5]
print(functools.reduce(lambda a,b:a+b,lst))
```

###Min element of list

```python
lst2 = [100, 87, 10, 39, -1, 0, 19, 20]
print(functools.reduce(lambda x,y:x if x<y else y, lst2))
```

```python
def tool_reduce(func, seq):
  res = seq[0]
  for el in seq[1:]:
    res=func(res,el)
    return res
```

```python
f = lambda x,y:x*y
```

```python
tool_reduce(f, [1, 10, 4, 5])
```

```python
n=5
```

```python
tool_reduce(f, range(1,n+1))
```
