```python

from abc import ABC, abstractmethod

class Account(ABC):                      # Abstraction
    bank_name = "MUBIS Bank"             # Class attribute (бүх объектод нийтлэг)

    def __init__(self, owner, balance=0):    # Constructor
        self.owner = owner               # Instance attribute
        self.__balance = balance         # Encapsulation (private)

    @property
    def balance(self):                   # Зөвхөн уншигдах шинж
        return self.__balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Дүн эерэг байх ёстой")
        self.__balance += amount

    def _withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("Үлдэгдэл хүрэлцэхгүй")
        self.__balance -= amount

    @abstractmethod
    def monthly_update(self): ...        # Дэд класс заавал хэрэгжүүлнэ

    def __str__(self):                   # Dunder (magic) method
        return f"{self.owner}: {self.balance:,.0f}₮"


class Savings(Account):                  # Inheritance
    def __init__(self, owner, balance=0, rate=0.01):
        super().__init__(owner, balance) # Эх классын constructor
        self.rate = rate

    def monthly_update(self):            # Polymorphism: хүү нэмнэ
        self.deposit(self.balance * self.rate)


class Checking(Account):
    FEE = 1000

    def monthly_update(self):            # Polymorphism: шимтгэл хасна
        self._withdraw(self.FEE)


accounts = [Savings("Бат", 100_000, 0.02), Checking("Сараа", 50_000)]
for acc in accounts:
    acc.monthly_update()                 # Ижил дуудлага, өөр үйлдэл
    print(acc)
# Бат: 102,000₮
# Сараа: 49,000₮

# acc.__balance = 0   → AttributeError биш ч жинхэнэ үлдэгдлийг өөрчлөхгүй
# Account("X")        → TypeError: abstract класс тул объект үүсгэх боломжгүй
```
